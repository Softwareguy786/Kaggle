```python
train['Arrival Delay in Minutes'] = train['Arrival Delay in Minutes'].fillna(train['Arrival Delay in Minutes'].median())
```

```python
test = pd.get_dummies(test, columns=['Gender','Customer Type','Type of Travel','Class'], drop_first=True)
test = test.drop(columns=['id'])
```

```python
train['satisfaction'] = train['satisfaction'].astype(int)
```
